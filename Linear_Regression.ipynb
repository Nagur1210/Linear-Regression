```python
numerical_columns = ['wheelbase','carlength','carwidth','carheight','curbweight','enginesize','boreratio','stroke','compressionratio','horsepower','peakrpm','citympg','highwaympg','price']
for col in numerical_columns:
    sns.histplot(df[col],kde = True)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    LL = Q1 - 1.5 * IQR
    UL = Q3 + 1.5 * IQR

    df[col] = df[col].clip(LL, UL)
```

```python
df.info()
```

```python
df=df.drop('car_ID',axis=1)
```

```python
df['brand'] = df['CarName'].str.split().str[0]
```

```python
df=df.drop('CarName',axis=1)
```

```python
df['avg_mpg'] = (df['citympg'] + df['highwaympg']) / 2
```

```python
df['car_volume'] =(
    df['carlength'] *
    df['carwidth'] *
    df['carheight'])
```

```python
df=df.drop(['citympg', 'highwaympg'], axis=1)
```

```python
df
```

```python
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap='coolwarm')
```

```python
x=df.drop('price',axis=1)
```

```python
y=df['price']
```

```python
x_encoded=pd.get_dummies(x,columns=['enginelocation','fueltype','doornumber','aspiration','carbody','drivewheel','drivewheel','enginetype','cylindernumber','fuelsystem','brand'],drop_first=True,dtype=int)
```

```python
x_encoded.info()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x_encoded,y,train_size=0.8,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc=StandardScaler()
```

```python
x_train=sc.fit_transform(x_train)
```

```python
x_test=sc.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression, lasso,Ridge
```

```python
LR=LinearRegression()
```

```python
LR.fit(x_train,y_train)
```

```python
y_pred=LR.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, root_mean_squared_error,mean_absolute_error,r2_score
```

```python
mean_squared_error(y_test,y_pred)
```

```python
root_mean_squared_error(y_test,y_pred)
```

```python
mean_absolute_error(y_test,y_pred)
```

```python
r2_score(y_test,y_pred)
```

```python
LR.coef_
```

```python
LR.intercept_
```

```python
LS=Lasso(alpha=8.0)
```

```python
LS.fit(x_train,y_train)
```

```python
y_pred_Lasso=LS.predict(x_test)
```

```python
root_mean_squared_error(y_test,y_pred_Lasso)
```

```python
mean_squared_error(y_test,y_pred_Lasso)
```

```python
mean_absolute_error(y_test,y_pred_Lasso)
```

```python
r2_score(y_test,y_pred_Lasso)
```

```python
LS.coef_
```

```python
LS.intercept_
```

```python
RR=Ridge(alpha=2.0)
RR.fit(x_train,y_train)
y_pred_Ridge=RR.predict(x_test)
```

```python
r2_score(y_test,y_pred_Ridge)
```

```python
mean_squared_error(y_test,y_pred_Ridge)
```

```python
root_mean_squared_error(y_test,y_pred_Ridge)
```

```python
mean_absolute_error(y_test,y_pred_Ridge)
```

```python
RR.coef_
```

```python
RR.intercept_
```

```python

comparison = pd.DataFrame({
    'Model': ['Linear Regression', 'Lasso Regression', 'Ridge Regression'],

    'MAE': [
        mean_absolute_error(y_test, y_pred),
        mean_absolute_error(y_test, y_pred_Lasso),
        mean_absolute_error(y_test, y_pred_Ridge)
    ],

    'MSE': [
        mean_squared_error(y_test, y_pred),
        mean_squared_error(y_test, y_pred_Lasso),
        mean_squared_error(y_test, y_pred_Ridge)
    ],

    'RMSE': [
        root_mean_squared_error(y_test, y_pred),
        root_mean_squared_error(y_test, y_pred_Lasso),
        root_mean_squared_error(y_test, y_pred_Ridge)
    ],

    'R2 Score': [
        r2_score(y_test, y_pred),
        r2_score(y_test, y_pred_Lasso),
        r2_score(y_test, y_pred_Ridge)
    ]
})

comparison
```

```python
coef_comparison = pd.DataFrame({
    'Feature': x_encoded.columns,
    'Linear Regression': LR.coef_,
    'Lasso Regression': LS.coef_,
    'Ridge Regression': RR.coef_
})

coef_comparison
```

All three models performed well in predicting car prices. Linear Regression gave an R² score of 0.9336, while Lasso achieved 0.9277. Ridge Regression performed slightly better with the highest R² score of 0.9359 and also had the lowest MSE and RMSE. Linear Regression, however, had the lowest MAE.
Lasso made some feature coefficients zero, which means it automatically removed features that were less useful for prediction. Overall, Ridge Regression performed the best. Since some of the features in the dataset are highly correlated with each other, Ridge helps reduce their influence without completely removing them, making the model more stable and better at predicting new data.
